# TP 1 : Les notes d'une classe


```python
notes = [12, 8, 15, 17, 9, 11, 14, 6, 13, 16]

# 1/ Nombre de notes,la moyenne arrondie,note minimale et note maximale
print("Nombre de notes :", len(notes))
print("Moyenne :", round(sum(notes) / len(notes), 2))
print("Note minimale :", min(notes))
print("Note maximale :", max(notes))
```

```python
# 2/ Notes superieures ou egales a 10
admis = [n for n in notes if n >= 10]
ajournes = len([n for n in notes if n < 10])

print("Notes des admis :", admis)
print("Nombre d'ajournes :", ajournes)
```

```python
# 3/ Tri croissant et trois meilleures notes
notes_triees = sorted(notes)
print("Notes triees :", notes_triees)
print("Trois meilleures notes :", sorted(notes, reverse=True)[:3])
```

```python
# 4/ Ajouter 2 points aux notes inferieures a 10,
# sans depasser 10
nouvelles_notes = [min(n + 2, 10) if n < 10 else n for n in notes]
print("Nouvelles notes :", nouvelles_notes)
```

```python
# 5/ Histogramme des notes
def histogramme(notes):
    tranches = [(0, 4), (5, 9), (10, 14), (15, 20)]

    for minimum, maximum in tranches:
        nombre = len([n for n in notes if minimum <= n <= maximum])
        print(f"{minimum}-{maximum} | {'*' * nombre}")

histogramme(notes)
```

```python
# 6/ Rang d'une note dans le classement decroissant
def rang(notes, note):
    classement = sorted(notes, reverse=True)
    return classement.index(note) + 1

print("Rang de la note 14 :", rang(notes, 14))
```

```python
# 7/ Bonus  fusionner sans doublons,

fusion = []
for n in notes + [15, 6, 18]:
    if n not in fusion:
        fusion.append(n)

print("Fusion sans doublons :", fusion)
```

```python
#3/Écrire la fonction bornes(villes)
def bornes(villes):
    latitudes = [ville[1] for ville in villes]
    longitudes = [ville[2] for ville in villes]

    return (
        min(latitudes),
        max(latitudes),
        min(longitudes),
        max(longitudes)
    )


lat_min, lat_max, lon_min, lon_max = bornes(villes)

print("Latitude minimale :", lat_min)
print("Latitude maximale :", lat_max)
print("Longitude minimale :", lon_min)
print("Longitude maximale :", lon_max)
```

```python
#7/ Bonus namedtuple et distance en kilomètres
from collections import namedtuple
from math import radians, sin, cos, asin, sqrt

Ville = namedtuple("Ville", ["nom", "latitude", "longitude"])

v1 = Ville("Antananarivo", -18.88, 47.51)
v2 = Ville("Toamasina", -18.15, 49.40)


def distance_km(v1, v2):
    lat1 = radians(v1.latitude)
    lon1 = radians(v1.longitude)
    lat2 = radians(v2.latitude)
    lon2 = radians(v2.longitude)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat / 2) ** 2
        + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    )

    c = 2 * asin(sqrt(a))
    return 6371 * c


print("Distance en km :", round(distance_km(v1, v2), 2))
```

```python
#4/Fonction vendre()
def vendre(stock, produit, quantite):
    infos = stock.get(produit)

    if infos is None:
        return (False, "Produit inconnu")
    if quantite <= 0:
        return (False, "Quantité invalide")
    if infos["quantite"] < quantite:
        return (False, "Quantité insuffisante")

    infos["quantite"] -= quantite
    return (True, "Vente réussie")

print("\n TESTS DE VENTE")
print("Vente réussie :", vendre(stock, "riz", 5))
print("Produit inconnu :", vendre(stock, "lait", 2))
print("Quantité trop grande :", vendre(stock, "huile", 100))
```

```python
#5 /Fonction ajouter_produit()
def ajouter_produit(stock, nom, prix, quantite):
    if nom in stock:
        return (False, "Produit déjà présent")
    if prix < 0 or quantite < 0:
        return (False, "Prix ou quantité invalide")

    stock[nom] = {"prix": prix, "quantite": quantite}
    return (True, "Produit ajouté")

print("\n TESTS D'AJOUT")
print(ajouter_produit(stock, "lait", 3500, 20))
print(ajouter_produit(stock, "riz", 3200, 10))
```

TP4 — Atelier décorateurs


```python
#1/ 
import time
from functools import wraps

def chronometrer(fonction):
    @wraps(fonction)
    def enveloppe(*args, **kwargs):
        debut = time.perf_counter()

        resultat = fonction(*args, **kwargs)

        fin = time.perf_counter()
        print(f"{fonction.__name__} a pris {fin - debut:.2f} s")

        return resultat

    return enveloppe


@chronometrer
def lente():
    time.sleep(1)


lente()
```

```python
#2/ 
from functools import wraps

def majuscules(fonction):
    @wraps(fonction)
    def enveloppe(*args, **kwargs):
        resultat = fonction(*args, **kwargs)
        return resultat.upper()

    return enveloppe


@majuscules
def saluer():
    return "bonjour"


print(saluer())
```

```python
#3/
from functools import wraps

def compter_appels(fonction):
    @wraps(fonction)
    def enveloppe(*args, **kwargs):
        enveloppe.appels += 1
        return fonction(*args, **kwargs)

    enveloppe.appels = 0
    return enveloppe


@compter_appels
def dire_bonjour():
    print("Bonjour !")


dire_bonjour()
dire_bonjour()
dire_bonjour()

print("Nombre d'appels :", dire_bonjour.appels)
```

```python
#4
from functools import wraps

def verifier_positifs(fonction):
    @wraps(fonction)
    def enveloppe(*args, **kwargs):
        for valeur in args:
            if isinstance(valeur, (int, float)) and valeur < 0:
                raise ValueError(
                    "Les valeurs doivent être positives."
                )

        return fonction(*args, **kwargs)

    return enveloppe


@verifier_positifs
def surface(longueur, largeur):
    return longueur * largeur


print(surface(5, 3))

try:
    print(surface(-5, 3))
except ValueError as erreur:
    print("Erreur :", erreur)
```

```python
#5/
import time
from functools import wraps

def chronometrer(unite="s"):
    def decorateur(fonction):
        @wraps(fonction)
        def enveloppe(*args, **kwargs):
            debut = time.perf_counter()

            resultat = fonction(*args, **kwargs)

            fin = time.perf_counter()
            duree = fin - debut

            if unite == "ms":
                duree *= 1000

            print(f"Durée : {duree:.2f} {unite}")
            return resultat

        return enveloppe

    return decorateur


@chronometrer(unite="ms")
def calcul():
    time.sleep(0.2)
    return "Terminé"


print(calcul())
```

```python
#6
from functools import wraps

def memoriser(fonction):
    cache = {}

    @wraps(fonction)
    def enveloppe(*args):
        if args in cache:
            return cache[args]

        resultat = fonction(*args)
        cache[args] = resultat
        return resultat

    return enveloppe


@memoriser
def fibonacci(n):
    if n < 2:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


print(fibonacci(30))

```

```python
#8/
from functools import wraps

class MiniFlask:
    def __init__(self):
        self.routes = {}
        self.gestionnaire_404 = None

    def route(self, chemin):
        def enregistrer(fonction):
            @wraps(fonction)
            def enveloppe(*args, **kwargs):
                return fonction(*args, **kwargs)

            if chemin in self.routes:
                raise ValueError("Cette URL existe déjà.")

            self.routes[chemin] = enveloppe
            return enveloppe

        return enregistrer

    def traiter(self, chemin):
        fonction = self.routes.get(chemin)

        if fonction:
            return fonction()

        if self.gestionnaire_404:
            return self.gestionnaire_404()

        return "404 : page introuvable"

    def lister_routes(self):
        for chemin in self.routes:
            print(chemin)

    def erreur_404(self, fonction):
        self.gestionnaire_404 = fonction
        return fonction


app = MiniFlask()


@app.route("/")
def accueil():
    return "Bonjour !"


@app.route("/contact")
def contact():
    return "Page contact"


@app.erreur_404
def page_introuvable():
    return "Page introuvable."


app.lister_routes()

print(app.traiter("/"))
print(app.traiter("/contact"))
print(app.traiter("/inconnu"))
```
